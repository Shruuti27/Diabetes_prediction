#### **I drop columns from the DataFrame**
I do it because they contain URLs, which are not needed for my analysis.
By removing these columns, I can focus on the more relevant data for my exploratory analysis (EDA).

```python
# Drop the columns with URLs (img_link and product_link)
df = df.drop(columns=['about_product', 'user_name', 'review_id',
    'review_title', 'review_content', 'img_link', 'product_link'])
print(df.head())
```

### **Remove currency symbols and commas, then convert to numeric** and **fill NaN values**

```python
# Remove currency symbols and commas, then convert 'discounted_price' and 'actual_price' to numeric
df['discounted_price'] = df['discounted_price'].replace({'₹': '', ',': ''}, regex=True).astype(float)
df['actual_price'] = df['actual_price'].replace({'₹': '', ',': ''}, regex=True).astype(float)

# Remove '%' and commas, then convert 'discount_percentage' to numeric
df['discount_percentage'] = df['discount_percentage'].replace({'₹': '', ',': '', '%': ''}, regex=True).astype(float)

# Remove any non-numeric characters from 'rating' (e.g., '|')
df['rating'] = df['rating'].replace({r'[^0-9.]': ''}, regex=True)

# Convert 'rating' to numeric values (float), handle errors by turning invalid entries into NaN
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Remove commas and convert 'rating_count' to numeric (int), first handle NaN values
df['rating_count'] = df['rating_count'].replace({',': ''}, regex=True)

# Convert to numeric, fill NaN values with 0
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce').fillna(0).astype(int)

print(df.head())
```

```python
# Remove duplicates based on all columns
df = df.drop_duplicates()
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")
```

### **Splitting `category` for better analysis, renaming new columns and dropping unnecessary columns**

By performing these splits, I can achieve a more granular view of the data, enabling more targeted analysis.

```python
# Split the 'category' column based on the '|' delimiter
category_split = df['category'].str.split('|', expand=True)

# Concatenate the split columns back into the original DataFrame
df = pd.concat([df, category_split], axis=1)

# Display the updated DataFrame with default column names (integer-based)
print(df)
```

```python
# Calculate Q1, Q3, and IQR
Q1 = df['actual_price'].quantile(0.25)
Q3 = df['actual_price'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Print the bounds
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

# Identify outliers
outliers = df[(df['actual_price'] < lower_bound) | (df['actual_price'] > upper_bound)]

# Display the outliers
print(f"Number of outliers: {outliers.shape[0]}")
outliers[['product_name', 'Category', 'Sub_category', 'Product', 'Product_type', 'actual_price']]
```

### **Data Visualization**

#### **Feature Engineering (Creating KPIs)**
Here, I'll create a few KPIs like the price-to-rating ratio, discount effectiveness, and product rating quality.

Example KPIs:

- Price-to-Rating Ratio: This will help assess whether expensive products have high ratings.
- Discount Effectiveness: Measures if higher discounts correlate with higher sales or ratings.

**Explanation:**

- **price_to_rating:** I divide the actual price by the rating to see if high-priced products have a good rating.
- **discount_effectiveness:** I multiply discount percentage by rating to see if bigger discounts lead to better ratings.
- **discount_to_price:** I calculate the ratio of discounted to actual price to see the discount's impact.

```python
# KPI 1: Price-to-Rating Ratio
df['price_to_rating'] = df['actual_price'] / df['rating']

# KPI 2: Discount Effectiveness (Discount percentage vs rating)
df['discount_effectiveness'] = df['discount_percentage'] * df['rating']

# KPI 3: Discount vs Price Comparison (Check if higher price means more discount)
df['discount_to_price'] = df['discounted_price'] / df['actual_price']

df[['product_name', 'actual_price', 'rating', 'price_to_rating', 'discount_percentage', 'discount_effectiveness', 'discount_to_price']].head()
```

### **Collaborative Filtering:**
- Build a user-item interaction matrix (**user_id** x **product_id**).
- Use **item-based** collaborative filtering to calculate product similarities using cosine similarity.
- Generate recommendations for each user based on the interaction matrix.

### **Let's build a system where we identify similar users based on their purchasing patterns**

**Create User-Product Interaction Matrix**

```python
# Aggregate interaction matrix by 'Product' for users
user_product_matrix = pd.pivot_table(
    df,
    index='user_id',
    columns='Product',
    aggfunc='size',
    fill_value=0
)

print("User-Product Interaction Matrix (first 5 rows):")
print(user_product_matrix.head())

# Visualize the interaction matrix sparsity
plt.figure(figsize=(12, 8))
sns.heatmap(user_product_matrix.iloc[:20, :20], cbar=False, cmap="coolwarm")
plt.title("User-Product Interaction Matrix (Sample)")
plt.xlabel("Products")
plt.ylabel("Users")
plt.show()
```

**Calculate User Similarity**

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compute user similarity using cosine similarity
user_similarity = cosine_similarity(user_product_matrix.values)

# Convert to DataFrame
user_similarity_df = pd.DataFrame(
    user_similarity,
    index=user_product_matrix.index,
    columns=user_product_matrix.index
)

print("User Similarity Matrix (first 5 rows and columns):")
print(user_similarity_df.iloc[:5, :5])

# Visualize the user similarity matrix
plt.figure(figsize=(14, 10))
sns.heatmap(
    user_similarity_df.iloc[:20, :20],
    cmap="Spectral",
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    linecolor="white"
)
plt.title("User Similarity Matrix Heatmap (Sample)", fontsize=18, fontweight="bold", pad=20)
plt.xlabel("Users", fontsize=14, fontweight="bold", labelpad=10)
plt.ylabel("Users", fontsize=14, fontweight="bold", labelpad=10)
plt.xticks(rotation=45, fontsize=10, ha='right')
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.show()
```

**Generate Recommendations for a Target User**

```python
def recommend_products_by_user_similarity(user_id, user_product_matrix, user_similarity_df, df, top_n=5):
    # Get similar users for the target user
    similar_users = user_similarity_df[user_id].sort_values(ascending=False).iloc[1:6].index.tolist()

    # Get products interacted by similar users
    similar_users_products = df[df['user_id'].isin(similar_users)]['product_id'].unique()

    # Get products already interacted by the target user
    target_user_products = df[df['user_id'] == user_id]['product_id'].unique()

    # Recommend products not already purchased by the target user
    recommendations = list(set(similar_users_products) - set(target_user_products))

    # Return the top N recommendations
    return recommendations[:top_n]

# Example user
user_id = df['user_id'].iloc[0]  # Replace with an actual user ID
recommendations = recommend_products_by_user_similarity(
    user_id, user_product_matrix, user_similarity_df, df
)

print(f"Recommendations for User '{user_id}':")
print(recommendations)
```

**Test Recommendations for a Single User**

```python
# Select a specific user for testing
test_user_id = df['user_id'].iloc[0]  # Replace with any valid user_id from the DataFrame

# Generate recommendations for the test user
test_recommendations = recommend_products_by_user_similarity(
    test_user_id,
    user_product_matrix,
    user_similarity_df,
    df,
    top_n=5  # Recommend top 5 products
)

# Output recommendations for the test user
print(f"Recommendations for User '{test_user_id}':")
print(test_recommendations)
```

**Generate Recommendations for All Users**

```python
# Function to generate recommendations for all users
def generate_recommendations_for_all_users(user_product_matrix, user_similarity_df, df, top_n=15):
    recommendations_dict = {}
    for user_id in user_product_matrix.index:  # Loop through all users
        recommendations = recommend_products_by_user_similarity(
            user_id,
            user_product_matrix,
            user_similarity_df,
            df,
            top_n
        )
        recommendations_dict[user_id] = recommendations
    return recommendations_dict

# Generate recommendations for all users
all_user_recommendations = generate_recommendations_for_all_users(
    user_product_matrix,
    user_similarity_df,
    df,
    top_n=15
)

# Display recommendations for the first 15 users
print("Recommendations for the first 15 users:")
for user_id, recs in list(all_user_recommendations.items())[:15]:
    print(f"User '{user_id}': {recs}")
```
